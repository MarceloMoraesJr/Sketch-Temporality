```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import pickle as pkl

sns.set_style("darkgrid")
```

```python
results_df = {"input_relative_coords": [], "output_relative_coords": [], "decoder_type": [], "test_acc": [], "seed": []}

for filename in os.listdir(f"../results/experiment4/"):

    with open(f"../results/experiment4/{filename}", "rb") as file:
        results = pkl.load(file)

        for key in results_df.keys():
            if key == "test_acc":
                value = results["test_results"][key]
            elif key not in results['pretraining_args'].keys():
                value = None
            else:
                value = results['pretraining_args'][key]
            
            results_df[key].append(value)

results_df = pd.DataFrame(results_df)
results_df = results_df.groupby(["decoder_type", "input_relative_coords", "output_relative_coords"], dropna=False)[['test_acc']].agg(['mean', 'sem'])
results_df[('test_acc', 'mean')] = results_df[('test_acc', 'mean')] * 100
results_df[('test_acc', 'mean')] = results_df[('test_acc', 'mean')].round(2)
results_df[('test_acc', 'sem')] = results_df[('test_acc', 'sem')] * 100
results_df[('test_acc', 'sem')] = results_df[('test_acc', 'sem')].round(2).map(lambda x: f"{x:05.2f}")
results_df
```
