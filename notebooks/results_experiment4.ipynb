```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import pickle as pkl

sns.set_style("darkgrid")
```

```python
results_df = {"input_relative_coords": [], "output_relative_coords": [], "decoder_type": [], "test_acc": [], "seed": []}

for filename in os.listdir(f"../../sketch-segmentation/results/experiment4/"):

    with open(f"../../sketch-segmentation/results/experiment4/{filename}", "rb") as file:
        results = pkl.load(file)

        for key in results_df.keys():
            if key == "test_acc":
                value = results["test_results"][key]
            elif key not in results['pretraining_args'].keys():
                value = None
            else:
                value = results['pretraining_args'][key]
            
            results_df[key].append(value)

results_df = pd.DataFrame(results_df)
results_df = results_df.groupby(["decoder_type", "input_relative_coords", "output_relative_coords"], dropna=False)[['test_acc']].agg(['mean', 'std'])
results_df[('test_acc', 'mean')] = results_df[('test_acc', 'mean')] * 100
results_df[('test_acc', 'mean')] = results_df[('test_acc', 'mean')].round(2)
results_df[('test_acc', 'std')] = results_df[('test_acc', 'std')] * 100
results_df[('test_acc', 'std')] = results_df[('test_acc', 'std')].round(2).map(lambda x: f"{x:05.2f}")
results_df
```

```python
import yaml

results_df = {"input_relative_coords": [], "output_relative_coords": [], "test_acc": [], "seed": []}

for filename in os.listdir(f"../configs/transfer/ffn/"):
    with open(f"../configs/transfer/ffn/{filename}", "r") as file:
        config = yaml.safe_load(file)

    with open(f"../{config['paths']['model_config']}", "r") as file:
        model_config = yaml.safe_load(file)

    results_df['input_relative_coords'].append(model_config['pe']['input_relative_coords'])
    results_df['output_relative_coords'].append(model_config['pe']['output_relative_coords'])
    results_df['seed'].append(model_config['training']['seed'])

    with open(f"../{config['paths']['results']}", "rb") as file:
        results = pkl.load(file)


    results_df["test_acc"].append(results["test_results"]["test_acc"])

results_df = pd.DataFrame(results_df)
results_df = results_df.groupby(["input_relative_coords", "output_relative_coords"])[['test_acc']].agg(['mean', 'std'])
results_df[('test_acc', 'mean')] = results_df[('test_acc', 'mean')] * 100
results_df[('test_acc', 'mean')] = results_df[('test_acc', 'mean')].round(2).map(lambda x: f"{x:05.2f}")
results_df[('test_acc', 'std')] = results_df[('test_acc', 'std')] * 100
results_df[('test_acc', 'std')] = results_df[('test_acc', 'std')].round(2).map(lambda x: f"{x:05.2f}")
results_df
```
