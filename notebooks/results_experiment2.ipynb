```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import pickle as pkl

sns.set_style("darkgrid")
```

```python
results_df = {"best_epoch": [], "best_step": [], "best_val_loss": [], "input_relative_coords": [], "output_relative_coords": [], "decoder_type": [], "test_l2_point": [], "test_l2_instance": [], "seed": []}

for filename in os.listdir(f"../../sketch_representations/checkpoints/experiment2/"):
    if not os.path.exists(f"../../sketch_representations/results/experiment2/{filename}.pkl"):
        continue

    checkpoint_df = pd.read_csv(f"../../sketch_representations/checkpoints/experiment2/{filename}/lightning_logs/version_0/metrics.csv")
    metrics = checkpoint_df.iloc[checkpoint_df['val_loss'].idxmin()]
    results_df["best_epoch"].append(metrics['epoch'].astype(int))
    results_df["best_step"].append(metrics['step'].astype(int))
    results_df["best_val_loss"].append(metrics['val_loss'])

    with open(f"../../sketch_representations/results/experiment2/{filename}.pkl", "rb") as file:
        results = pkl.load(file)

        for key in results_df.keys():
            if key in ["run", "best_epoch", "best_step", "best_val_loss"]: continue
            results_df[key].append(results["test_results"][key] if key in ["test_l2_point", "test_l2_instance"] else results["args"][key])

results_df = pd.DataFrame(results_df)
results_df.groupby(["decoder_type", "input_relative_coords", "output_relative_coords"])[['test_l2_point', 'test_l2_instance', 'best_epoch', 'best_step', 'best_val_loss']].agg(['mean', 'sem'])
```

```python
results_df = {"best_epoch": [], "best_step": [], "best_val_loss": [], "relative_coords": [], "pen_state": [], "stroke_embedding": [], "sketch_pos": [], "stroke_pos": [], "test_p_metric": [], "test_c_metric": [], "seed": []}

for filename in os.listdir(f"../checkpoints/experiment3/"):
    if not os.path.exists(f"../results/experiment3/{filename}.pkl"):
        continue

    checkpoint_df = pd.read_csv(f"../checkpoints/experiment3/{filename}/lightning_logs/version_0/metrics.csv")
    metrics = checkpoint_df.iloc[checkpoint_df['val_loss'].idxmin()]
    results_df["best_epoch"].append(metrics['epoch'].astype(int))
    results_df["best_step"].append(metrics['step'].astype(int))
    results_df["best_val_loss"].append(metrics['val_loss'])

    with open(f"../results/experiment3/{filename}.pkl", "rb") as file:
        results = pkl.load(file)

        for key in results_df.keys() - {"dataset"}:
            if key in ["run", "best_epoch", "best_step", "best_val_loss"]: continue
            results_df[key].append(results["test_results"][key] if key in ["test_p_metric", "test_c_metric"] else results["args"][key])

results_df = pd.DataFrame(results_df)
results_df.groupby(["relative_coords", "pen_state", "stroke_embedding", "sketch_pos", "stroke_pos"])[['test_p_metric', 'test_c_metric', 'best_epoch', 'best_step', 'best_val_loss']].agg(['mean', 'sem'])
```
