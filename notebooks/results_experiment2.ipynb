```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import pickle as pkl

sns.set_style("darkgrid")
```

```python
results_df = {"best_epoch": [], "best_step": [], "best_val_loss": [], "input_relative_coords": [], "output_relative_coords": [], "decoder_type": [], "test_l2_point": [], "test_l2_instance": [], "seed": []}

for filename in os.listdir(f"../checkpoints/experiment2/"):
    if not os.path.exists(f"../results/experiment2/{filename}.pkl"):
        continue

    checkpoint_df = pd.read_csv(f"../checkpoints/experiment2/{filename}/lightning_logs/version_0/metrics.csv")
    metrics = checkpoint_df.iloc[checkpoint_df['val_loss'].idxmin()]
    results_df["best_epoch"].append(metrics['epoch'].astype(int))
    results_df["best_step"].append(metrics['step'].astype(int))
    results_df["best_val_loss"].append(metrics['val_loss'])

    with open(f"../results/experiment2/{filename}.pkl", "rb") as file:
        results = pkl.load(file)

        for key in results_df.keys():
            if key in ["run", "best_epoch", "best_step", "best_val_loss"]: continue
            results_df[key].append(results["test_results"][key] if key in ["test_l2_point", "test_l2_instance"] else results["args"][key])

results_df = pd.DataFrame(results_df)
results_df = results_df.groupby(["decoder_type", "input_relative_coords", "output_relative_coords"])[['test_l2_point', 'test_l2_instance', 'best_epoch', 'best_step', 'best_val_loss']].agg(['mean', 'std'])
results_df[('test_l2_point', 'mean')] = results_df[('test_l2_point', 'mean')] * 100
results_df[('test_l2_point', 'mean')] = results_df[('test_l2_point', 'mean')].round(3).map(lambda x: f"{x:06.3f}")
results_df[('test_l2_point', 'std')] = results_df[('test_l2_point', 'std')] * 100
results_df[('test_l2_point', 'std')] = results_df[('test_l2_point', 'std')].round(3).map(lambda x: f"{x:06.3f}")
results_df
```

```python
import yaml
from pathlib import Path

results_df = {"input_relative_coords": [], "pen_state": [], "stroke_embedding": [], "stroke_pos": [], "sketch_pos": [], "test_l2_point": [], "seed": []}

for filename in os.listdir(f"../configs/pe/reconstruction/"):
    if filename == "base_config.yaml": continue
    with open(f"../configs/pe/reconstruction/{filename}", "r") as file:
        config = yaml.safe_load(file)
    results_path = Path("../").joinpath(config['paths']['results'])
    if not results_path.exists(): continue



    results_df['input_relative_coords'].append(config['pe']['input_relative_coords'])
    results_df['pen_state'].append(config['pe']['pen_state'])
    results_df['stroke_embedding'].append(config['pe']['stroke_embedding'])
    results_df['stroke_pos'].append(config['pe']['stroke_pos'])
    results_df['sketch_pos'].append(config['pe']['sketch_pos'])
    results_df['seed'].append(config['training']['seed'])
 
    with open(results_path, "rb") as file:
        results = pkl.load(file)


    results_df["test_l2_point"].append(results["test_results"]["test_l2_point"])

results_df = pd.DataFrame(results_df)
results_df = results_df.groupby(["input_relative_coords", "pen_state", "stroke_embedding", "stroke_pos", "sketch_pos"])[['test_l2_point']].agg(['mean', 'std'])
results_df[('test_l2_point', 'mean')] = results_df[('test_l2_point', 'mean')] * 100
results_df[('test_l2_point', 'mean')] = results_df[('test_l2_point', 'mean')].round(3).map(lambda x: f"{x:06.3f}")
results_df[('test_l2_point', 'std')] = results_df[('test_l2_point', 'std')] * 100
results_df[('test_l2_point', 'std')] = results_df[('test_l2_point', 'std')].round(3).map(lambda x: f"{x:06.3f}")
results_df
```
